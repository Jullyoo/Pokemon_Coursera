PRIMEIRAS ANALISES

```python
# Agrupando os Pokémons por Tipo 1 e calculando a média de ataque de cada tipo, ordenando do maior para o menor.

type1 = df.groupby('type_1')['attack'].mean().sort_values(ascending=False)

type1
```

```python
# verificando a média de cada estatística por tipo de Pokémon

type1 = df.groupby("type_1")[["hp", "attack", "defense", "sp_attack", "sp_defense", "speed"]].mean()
type1.sort_values("attack", ascending=False)
```

```python
bst = df.groupby("type_1")["base_stats"].mean().sort_values(ascending=False)

bst
```

```python
# Poder por geração

gen_power = df.groupby('gen')['base_stats'].mean().sort_values(ascending=False)

gen_power
```

```python
# Quantidade de Pokémons por geração

gen_count = df.groupby('gen').size()

gen_count
```

```python
# Poder x Captura

df["base_stats"].corr(df["capturing_rate"])

plt.scatter(df["capturing_rate"], df["base_stats"])

plt.xlabel("Taxa de Captura")
plt.ylabel("Poder Base")
plt.title("Relação entre Poder Base e Taxa de Captura")
plt.show()
```

```python
# Pokemons lendários x Não lendários - BST

df.groupby("is_legendary")["base_stats"].agg(['count', 'mean', 'median', 'min', 'max'])
```

```python
# Pokemons lendários x Não lendários - Taxa de Captura

df.groupby("is_legendary")["capturing_rate"].agg(['count', 'mean', 'median', 'min', 'max'])
```

```python
# Pokemons lendários x Não lendários - Estatísticas

df.groupby("is_legendary")[
    ["hp", "attack", "defense",
     "sp_attack", "sp_defense", "speed",
     "base_stats"]
].mean()
```

```python
# Analise por tipos

type_analysis = (
    df.groupby("type_1")
    .agg(
        pokemon_count=("name", "count"),
        avg_hp=("hp", "mean"),
        avg_attack=("attack", "mean"),
        avg_defense=("defense", "mean"),
        avg_speed=("speed", "mean"),
        avg_bst=("base_stats", "mean")
    )
    .sort_values("avg_bst", ascending=False)
)

type_analysis
```

```python
# Equilibrio de atributos

stats = ["hp", "attack", "defense", "sp_attack", "sp_defense", "speed"]

df["stats_std"] = df[stats].std(axis=1)

df[["name","base_stats", "stats_std"]].sort_values("stats_std").head(10) # Quanto menor o stats_std, mais equilibrado é o Pokémon.
```

```python
df[["name","base_stats", "stats_std"]].sort_values("stats_std", ascending=False).head(10) # Quanto maior o stats_std, mais desequilibrado é o Pokémon.
```
